```python
import numpy as np
import matplotlib.pyplot as plt

class Source:
    def __init__(self, Y=10000.0):
        self.Y = Y 

    def generate_photons(self, energy):
        mean_photons = self.Y * energy
        total_photons = np.random.poisson(mean_photons)
        batch_size = 100
        current_batch = []
        print(f"Сгенерировано событие с энергией {energy} МэВ. Ожидалось ~{mean_photons:.0f}, выпало: {total_photons} фотонов.")
        
        for _ in range(total_photons):
            cos_theta = np.random.uniform(-1, 1)
            sin_theta = np.sqrt(1 - cos_theta**2)
            phi = np.random.uniform(0, 2 * np.pi)
            nx = sin_theta * np.cos(phi)
            ny = sin_theta * np.sin(phi)
            nz = cos_theta
            current_batch.append((nx, ny, nz))

            if len(current_batch) == batch_size:
                yield current_batch
                current_batch = [] 


        if current_batch:
            yield current_batch

source = Source(Y=10000)
energy = 1.0 # МэВ


batch_counter = 1
total_photons_received = 0

for batch in source.generate_photons(energy):
    batch_len = len(batch)
    total_photons_received += batch_len
    
    # Первые 3 точки
    if batch_counter <= 3:
        print(f"Пачка №{batch_counter}: получено {batch_len} фотонов. Первый вектор: {batch[0]}")
    
    batch_counter += 1

print(f"\nИтого получено фотонов за все время: {total_photons_received}")
```
