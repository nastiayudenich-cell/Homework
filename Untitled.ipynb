```python
import numpy as np
import matplotlib.pyplot as plt
import vegas

lamda, N = 2, 100_000


rng = np.random.default_rng(1)
U = rng.uniform(0, 1, N)
X = -np.log(1-U) / lamda


x_vals = np.linspace(0, 6, 500)
pdf_vals = lamda * np.exp(-lamda * x_vals)
plt.plot(x_vals, pdf_vals, color="red", linewidth=2, label="Аналитическая плотность")

# гистограмма
plt.hist(X, bins=60, density=True, alpha=0.6)
plt.xlabel("x")
plt.ylabel("f")
plt.legend()
plt.show()

# Выборочные среднее и дисперсия
sample_mean = X.mean()
sample_var = X.var(ddof=1)
print(f"Выборочное среднее:   {sample_mean:.4f}")
print(f"Выборочная дисперсия: {sample_var:.4f}")


integ = vegas.Integrator([[0, 1]])
# E[X]
vegas_mean = integ(lambda u: -np.log(u[0]) / lamda, nitn=5, neval=10_000)

#  E[X^2]
vegas_second = integ(lambda u: (-np.log(u[0]) / lamda)**2, nitn=5, neval=10_000)

# D[X] = E[X^2] - (E[X])^2
vegas_var = vegas_second.mean - vegas_mean.mean**2

print(f"VEGAS среднее:        {vegas_mean.mean:.4f} (погрешность: {vegas_mean.sdev:.6f})")
print(f"VEGAS дисперсия:      {vegas_var:.4f}")

```
